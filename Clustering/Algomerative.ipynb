```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset = pd.read_csv('Mall_Customers.csv')
dataset
```

```python
X = dataset.iloc[:, 3:5].values
X
```

```python
import scipy.cluster.hierarchy as sch
import matplotlib.pyplot as plt

dendrogram = sch.dendrogram(
    sch.linkage(X, method='ward')
)

plt.title('Dendrogram')
plt.xlabel('Customers')
plt.ylabel('Euclidean distances')
plt.show()
```

```python
from sklearn.cluster import AgglomerativeClustering

# Create the Agglomerative Clustering model
clusmodel = AgglomerativeClustering(n_clusters=5)

# Fit the model and predict the cluster for each customer
label = clusmodel.fit_predict(X)
label
```

```python
# Create a new DataFrame from the original dataset
supervised = pd.DataFrame(dataset)
# Add the cluster labels as a new column
supervised['Cluster_group'] = label
# Display the final dataset
supervised
```

```python
import seaborn as sns
facet = sns.lmplot(data=supervised, x=supervised.columns[3], y=supervised.columns[4], hue=supervised.columns[5],
                   fit_reg=False,legend_out=True)
```
