```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset = pd.read_csv('insurance.csv')
dataset
```

```python
dataset=pd.get_dummies(dataset,drop_first=True)
dataset
```

```python
indep=dataset[['age', 'bmi', 'children','sex_male', 'smoker_yes','region_northwest','region_southeast','region_southwest']]
dep=dataset['charges']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(indep, dep, test_size = 1/3, random_state = 0)
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

```python
from sklearn.ensemble import AdaBoostRegressor
# regr = AdaBoostRegressor(random_state=0, n_estimators=100)
# regr.fit(X, y)
from sklearn.model_selection import GridSearchCV
param_grid = {'estimator':[None,'object'],
              'learning_rate': [1.0, 1.5, 2.0,2.5],
'random_state': [None, 0, 10, 100, 1000]}
```

```python
grid = GridSearchCV(AdaBoostRegressor(), param_grid, refit = True, verbose=3,n_jobs=-1)
grid.fit(X_train, y_train)
```

```python
re=grid.cv_results_
print("The R_score value for best parameter {}:".format(grid.best_params_))
```

```python
table=pd.DataFrame.from_dict(re)
table
```
